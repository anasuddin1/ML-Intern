## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: a content reviewer with limited time must choose which pages to look at first out of thousands.
Action: review the top-K pages in the ranked queue and refresh, expand, prune, or monitor them, guided by reason codes.
Cost of a wrong call: a false positive wastes reviewer hours on a healthy page; a false negative means a genuinely declining page keeps losing traffic unnoticed. Since review capacity is small (e.g. 50 pages), ranking quality at the top of the list (precision@K) matters more than overall accuracy.
Why ML helps: a fixed rule can't weigh many weak signals together (age, CTR, position, engagement, trend); a model can, and the starter run already showed the lift is real.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total = len(df)
declining = (df["trend_direction"] == "down").mean() * 100
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()

print(f"Total pages in starter data: {total:,}")
print(f"Pages currently declining: {declining:.1f}%")
print(f"Stale but visible pages (no update in 180+ days, 500+ impressions): {stale_visible:,}")
```

What these numbers say: 30,000 pages, and 54.2% (~16,000) are currently trending down — far more than any reviewer can check by hand, so a ranked queue is worth building. Interestingly, the classic "stale but visible" rule catches only 17 pages here, which shows a single hand-written rule misses most of the problem — more evidence that combining many signals (with a model) is needed.
